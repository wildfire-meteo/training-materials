## When does a plume rise?

A plume rises when it is buoyant, i.e. when a parcel of plume air is lighter than its surroundings. In practice, when we mean "lighter", we will usually say "warmer": a cubic metre of warm air weighs less than a cubic metre of cold air. So a parcel of air that is warmer than the air around it (at the same height, i.e. on the same isobar) will rise, and a parcel that is colder will sink. ## Stability and potential temperature

If you imagine lifting a parcel vertically through a "layer" of the atmosphere, and the parcel is positively buoyant (i.e. warmer than the atmosphere around it) the whole way, then that parcel is very likely to rise through the layer on its own. This is because it's always buoyant and thus can develop a vertical speed: Warm air goes up. Such a layer is called "unstable".

But how do you know if a parcel with a temperature at one height will rise? You need to know its temperature at the next height. To figure that out, you have to again consider that a rising parcel must expand to a lower pressure, and so it loses temperature. If it does this without losing energy (we will explain in more detail what this means on the first Monday), we call it *dry adiabatic* ascent, and the parcel follows one of the red lines in the diagram: the *dry adiabats*. Along a dry adiabat, the so-called *potential temperature* stays the same (this is the temperature where the dry adiabat crosses 1000 hPa). Such dry adiabatic descent is a pretty good model of how parcels actually rise, so we will use it to determine if a parcel is likely to rise and become a plume.

Lift a parcel of air from the surface with the slider on the plot below. It keeps its potential temperature, so it follows the dry adiabat. Up to which height is it warmer than its surroundings, i.e. positively buoyant?

````python
intro.dry_ascent_figure(sounding)
````

The parcel is warmer than its surroundings up to about 1800 m; above that, it is colder, so it stops rising there. The reason the parcel at the surface can rise freely on this day at Martorell, is thus because the surface is warmer than the atmosphere above it (the red line drops fast at first), and because the atmosphere then itself follows the dry adiabat (tick red line follows the thin red line), the parcel can keep going until the atmosphere's temperature line starts to tilt to the right again. This structure in the atmosphere is called the **atmospheric boundary layer**. ## Moisture, dew-point temperature and isohumes

The Martorell picture also shows a small clouds in white. Clouds are condensated water vapour, but our diagram doesn't show any moisture yet. We do that by plotting a second line from the forecast, namely the vertical variation of the *dew point temperature*.

````python
intro.dewpoint_figure(sounding)
````

What is the dew-point temperature? If you have water vapour in air, it will condense into droplets if you cool the air down enough. This is why you will often see droplets on the outside of a glass of cold water: The air next to the glass is cooled to its dew-point and some of the water in it condenses. How much you need to cool it depends on how much water vapour there is, which we measure as the *specific humidity*: the grams of water vapour found in a kilogram of air. From the specific humidity, you can calculate the temperature to which you must cool the air down for its vapour to condense. That temperature is called the dew point.

All you need to remember, is that the dew-point temperature line is a way of expressing how much water there is in the air, in a plot which can only show temperatures.

The dew point (blue) is to the left of the temperature (red) everywhere: nowhere is the air cool enough for its water vapour to condense, so there are no clouds in this sounding. To get a cloud, you would need to cool the air down more, *or* give a parcel rising through this atmosphere more water than there is in its surroundings.

So how do we track the amount of moisture in a rising parcel? We will just assume it just always stays the same. So we add lines to the skew-T diagram with a constant specific humidity, called *isohumes*. They are blue upwards-sloping lines, labelled in grams of water vapour per kilogram of air. Don't confuse them with the isotherms!

````python
intro.isohumes_figure(sounding)
````

Look at the dew point in the boundary layer, below about 1800 m: it runs nearly along an isohume. The boundary layer has about the same specific humidity at all heights. As we will discuss on Monday morning, this is because it is "well-mixed"; the rising parcels in the boundary layer can supply the same amount of moisture from the surface to all levels within the layer.

What is an isohume really? Exactly how they are calculated is a bit complex, so we skip it, but you can just think of them as the dew-point lines of rising parcels. So where a parcel's isohume meets its dry adiabat, its dew-point is equal to its temperature, i.e. the parcel is cool enough for its moisture to condense. The height at which this happens for idealised parcels like those we look at here is called the *lifting condensation level* (LCL). It is a reasonable model of the cloud base.

## Latent heat and moist adiabatic rise

Above the LCL, a parcel's dew point and temperature are the same, because the parcel is a cloud. But as it rises further and cools more, the rest of its water vapour slowly condenses. Condensation releases heat, called *latent heat*: this is the same heat that evaporating water takes away, e.g. the effect of sweating cooling you down. Therefore, the temperature of a cloudy parcel reduces with height less fast than in dry plumes. In such a process, another temperature (the equivalent potential temperature) is conserved.

Therefore, we also usually plot the lines of these ascents in the Skew T diagram: they are called *moist adiabats* (or sometimes saturated adiabats), and they are green in the plot below. Because a parcel on a moist adiabat cools more slowly, it can become warmer than its environment again, and therefore positively buoyant.

**Example.** Below, the surface parcel rises along its dry adiabat to the LCL, and then it continues along a moist adiabat above it (the black line, which is parallel to the green lines). For effect, we have also drawn a grey line showing where the parcel would go if there were no condensation.

````python
intro.moist_ascent_figure(sounding)
````

Without condensation, the parcel quickly becomes much colder than its surroundings, and would stop. But on the moist adiabat, it stays just warmer than its surroundings up to about 4 km (hover over the lines to check). That means that the surface parcel in this case might reach 4 km, instead of 2 km it would reach in a dry world: That is twice the height.

## When will the fire produce a pyrocloud?

From the discussion above, you will notice that he simplest condition for pyrocloud formation is that plume air just above the fire must be so warm that it reaches the LCL while being warmer than its environment all the way there. Of course, a wildfire heats the plume air quite a lot, so surface parcels above a wildfire don't just start with the same temperature as the forecast, but with a higher temperature. This can make the plume top change quite dramatically. Use the slider on the plot below to make the plume's surface air warmer, and read off both the height and the temperature of its cloud base. Can you expect a deep pyrocloud in this situation?

````python
intro.pyrocloud_figure(sounding)
````

Without extra heat, the surface air just about reaches its cloud base, at about 1800 m. As the plot says: The situation is on the edge. But at about 1 °C of extra heating from the fire, you could reach a cloud base which is clearly warmer than the surroundings and form a deeper pyrocumulus cloud. This is the type of calculation that is often performed to assess the risk of pyroconvection. As we will discuss in the training, it's also too simple: Real plumes mix with the air around them as they rise, which slows them down.
