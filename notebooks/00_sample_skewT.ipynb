# Sample: a skew-T diagram

This page checks that the interactive figures work in your browser.

Below is a sounding from Madrid, 24 June 1998, at noon (12 UTC): the **temperature** in red and the **dew point** in blue.

- Hover over the red or blue line to read its temperature and pressure.
- Click a name in the legend to show or hide those lines.
- Drag a box to zoom in; double-click to zoom out again.

```python
from odet_meteo import skewt, soundings

sounding = soundings.load_sounding("madrid_1998062412")
skewt.skewt_figure(sounding)
```

## A slider

Move the slider to follow a dry adiabat: the line along which air cools as it rises without forming cloud.

```python
skewt.add_dry_adiabat_slider(skewt.skewt_figure(sounding))
```
